## 5.5 Overall Model Performance Analysis

The performance of five classification algorithms was evaluated using Accuracy, AUC, Precision, Recall, F1 Score and Matthews Correlation Coefficient.

The following observations were obtained:

1. **Logistic Regression:** Achieved the highest AUC (0.7376) and MCC (0.2429), demonstrating the best overall discrimination and correlation with actual outcomes.

2. **Decision Tree:** Achieved relatively low performance, with an AUC of 0.5657 and MCC of 0.1231.

3. **K-Nearest Neighbors:** Achieved an accuracy of 87.62%, but its recall was only 4.81%, indicating poor identification of actual subscribers.

4. **Gaussian Naive Bayes:** Achieved the highest Recall (40.38%) and F1 Score (0.3281), making it comparatively effective at identifying customers who subscribed.

5. **Support Vector Machine:** Achieved the highest Accuracy (88.84%) and Precision (56%), but its Recall was relatively low at 13.46%.

### Overall Winner: Logistic Regression

Logistic Regression is selected as the overall winner because it achieved the highest AUC and MCC among the five models. However, Gaussian Naive Bayes demonstrated better performance in identifying actual subscribers, as reflected by its higher Recall and F1 Score.

The results also demonstrate that Accuracy alone is insufficient for evaluating classification models on an imbalanced dataset.
